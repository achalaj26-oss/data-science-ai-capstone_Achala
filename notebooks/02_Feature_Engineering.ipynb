## Customer Sampling for Model Development

The original American Express dataset contains **458,913 unique customers** and **5.5 million+ monthly observations**.

Due to the computational and memory requirements of processing the complete dataset in the Colab environment, a sample of **100,000 unique customers** is selected for model development.

**Stratified sampling** is used based on the target variable to preserve the original default/non-default class distribution.

The selected sample will be used for:
- Feature engineering
- Model development
- Model evaluation
- Explainable AI (SHAP) analysis

A fixed random seed (`random_state=42`) is used to ensure reproducibility.

```python

LABEL_PATH = "/content/drive/MyDrive/AMEX_Credit_Risk_Capstone_Project/data/raw/train_labels.csv"

labels = pd.read_csv(LABEL_PATH)

# Select 100,000 customers while preserving the original target distribution

sampled_labels = (
    labels
    .groupby("target", group_keys=False)
    .sample(
        frac=100000 / len(labels),
        random_state=42
    )
)

print("Selected customers:", len(sampled_labels))

print("\nTarget distribution:")
print(sampled_labels["target"].value_counts())

print("\nTarget percentage:")
print(
    sampled_labels["target"]
    .value_counts(normalize=True)
    .mul(100)
)
```

```python
def create_customer_features(
    input_file,
    output_file,
    batch_size=20
):

    print("=" * 70)
    print("Processing:", os.path.basename(input_file))

    # ---------------------------------------------------------
    # 1. Read only the column names
    # ---------------------------------------------------------
    columns = pd.read_csv(
        input_file,
        nrows=0
    ).columns.tolist()

    id_col = "customer_ID"
    date_col = "S_2"

    # Numerical candidate columns
    excluded = [id_col, date_col, "D_63", "D_64"]

    numeric_cols = [
        c for c in columns
        if c not in excluded
    ]

    print("Total columns:", len(columns))
    print("Numerical features:", len(numeric_cols))

    # ---------------------------------------------------------
    # 2. Create customer index
    # ---------------------------------------------------------
    ids = pd.read_csv(
        input_file,
        usecols=[id_col]
    )

    customer_ids = ids[id_col].drop_duplicates().reset_index(drop=True)

    print("Unique customers:", len(customer_ids))

    # Final feature dataframe
    customer_features = pd.DataFrame({
        id_col: customer_ids
    })

    del ids
    del customer_ids

    # ---------------------------------------------------------
    # 3. Process numerical features in batches
    # ---------------------------------------------------------
    for start in range(0, len(numeric_cols), batch_size):

        batch_cols = numeric_cols[start:start + batch_size]

        print("\nProcessing features:",
              start + 1,
              "to",
              start + len(batch_cols),
              "of",
              len(numeric_cols))

        usecols = [id_col, date_col] + batch_cols

        df = pd.read_csv(
            input_file,
            usecols=usecols
        )

        # Ensure date is datetime
        df[date_col] = pd.to_datetime(
            df[date_col],
            errors="coerce"
        )

        # Sort so "last" really means latest observation
        df = df.sort_values(
            [id_col, date_col]
        )

        # -----------------------------------------------------
        # Aggregate
        # -----------------------------------------------------
        grouped = df.groupby(id_col, sort=False)

        features = grouped[batch_cols].agg(
            ["mean", "std", "min", "max", "last"]
        )

        # Flatten column names
        features.columns = [
            f"{col}_{stat}"
            for col, stat in features.columns
        ]

        features = features.reset_index()

        # -----------------------------------------------------
        # Merge with customer feature table
        # -----------------------------------------------------
        customer_features = customer_features.merge(
            features,
            on=id_col,
            how="left",
            validate="one_to_one"
        )

        del df
        del grouped
        del features

        print(
            "Current feature table:",
            customer_features.shape
        )

    # ---------------------------------------------------------
    # 4. Add observation count
    # ---------------------------------------------------------
    counts = pd.read_csv(
        input_file,
        usecols=[id_col]
    )[id_col].value_counts()

    customer_features["observation_count"] = (
        customer_features[id_col].map(counts)
    )

    # ---------------------------------------------------------
    # 5. Save
    # ---------------------------------------------------------
    print("\nSaving:", output_file)

    customer_features.to_csv(
        output_file,
        index=False
    )

    print("Saved successfully.")
    print("Final shape:", customer_features.shape)

    return customer_features
```

```python

# ---------------------------------------------------------
# Paths
# ---------------------------------------------------------

source_file = os.path.join(
    PROCESSED_PATH,
    "train_data_100k_no_90pct_missing_v2.csv"
)

train_ids_file = os.path.join(
    PROCESSED_PATH,
    "train_customer_ids.csv"
)

val_ids_file = os.path.join(
    PROCESSED_PATH,
    "validation_customer_ids.csv"
)

test_ids_file = os.path.join(
    PROCESSED_PATH,
    "test_customer_ids.csv"
)

train_output = os.path.join(
    PROCESSED_PATH,
    "train_monthly_70k_v2.csv"
)

val_output = os.path.join(
    PROCESSED_PATH,
    "validation_monthly_15k_v2.csv"
)

test_output = os.path.join(
    PROCESSED_PATH,
    "test_monthly_15k_v2.csv"
)

# ---------------------------------------------------------
# Load ONLY customer IDs
# ---------------------------------------------------------

train_ids = set(
    pd.read_csv(
        train_ids_file,
        usecols=["customer_ID"]
    )["customer_ID"]
)

val_ids = set(
    pd.read_csv(
        val_ids_file,
        usecols=["customer_ID"]
    )["customer_ID"]
)

test_ids = set(
    pd.read_csv(
        test_ids_file,
        usecols=["customer_ID"]
    )["customer_ID"]
)

print("Train customers:", len(train_ids))
print("Validation customers:", len(val_ids))
print("Test customers:", len(test_ids))

# ---------------------------------------------------------
# Make sure there is no overlap
# ---------------------------------------------------------

print("\nChecking customer overlap...")

print(
    "Train ∩ Validation:",
    len(train_ids & val_ids)
)

print(
    "Train ∩ Test:",
    len(train_ids & test_ids)
)

print(
    "Validation ∩ Test:",
    len(val_ids & test_ids)
)

# ---------------------------------------------------------
# Remove old output files if they exist
# ---------------------------------------------------------

for file in [
    train_output,
    val_output,
    test_output
]:
    if os.path.exists(file):
        os.remove(file)

# ---------------------------------------------------------
# Process source file in chunks
# ---------------------------------------------------------

chunk_size = 100_000

train_rows = 0
val_rows = 0
test_rows = 0

first_train = True
first_val = True
first_test = True

chunk_number = 0

for chunk in pd.read_csv(
    source_file,
    chunksize=chunk_size,
    low_memory=False
):

    chunk_number += 1

    # ---------------------------------------------
    # Train
    # ---------------------------------------------

    train_chunk = chunk[
        chunk["customer_ID"].isin(train_ids)
    ]

    if len(train_chunk) > 0:

        train_chunk.to_csv(
            train_output,
            mode="w" if first_train else "a",
            header=first_train,
            index=False
        )

        first_train = False
        train_rows += len(train_chunk)

    # ---------------------------------------------
    # Validation
    # ---------------------------------------------

    val_chunk = chunk[
        chunk["customer_ID"].isin(val_ids)
    ]

    if len(val_chunk) > 0:

        val_chunk.to_csv(
            val_output,
            mode="w" if first_val else "a",
            header=first_val,
            index=False
        )

        first_val = False
        val_rows += len(val_chunk)

    # ---------------------------------------------
    # Test
    # ---------------------------------------------

    test_chunk = chunk[
        chunk["customer_ID"].isin(test_ids)
    ]

    if len(test_chunk) > 0:

        test_chunk.to_csv(
            test_output,
            mode="w" if first_test else "a",
            header=first_test,
            index=False
        )

        first_test = False
        test_rows += len(test_chunk)

    print(
        f"Chunk {chunk_number} processed | "
        f"Train: {train_rows:,} | "
        f"Validation: {val_rows:,} | "
        f"Test: {test_rows:,}"
    )

print("\nFinished!")
print("Train rows:", train_rows)
print("Validation rows:", val_rows)
print("Test rows:", test_rows)
print("Total rows:", train_rows + val_rows + test_rows)
```
