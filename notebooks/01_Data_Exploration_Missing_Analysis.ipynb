```python
# All predictor columns
all_columns = pd.read_csv(
    train_path,
    nrows=0
).columns.tolist()

feature_columns = [
    col for col in all_columns
    if col not in ["customer_ID", "S_2"]
]

print("Predictor variables:", len(feature_columns))
```

```python
missing_counts = pd.Series(
    0,
    index=feature_columns,
    dtype="int64"
)

total_rows = 0

for i, chunk in enumerate(
    pd.read_csv(
        train_path,
        usecols=feature_columns,
        chunksize=100000
    )
):

    missing_counts += chunk.isna().sum()
    total_rows += len(chunk)

    if (i + 1) % 10 == 0:
        print(
            f"Processed {total_rows:,} rows"
        )

missing_profile = pd.DataFrame({
    "feature": missing_counts.index,
    "missing_count": missing_counts.values,
})

missing_profile["missing_pct"] = (
    missing_profile["missing_count"]
    / total_rows
) * 100

missing_profile = missing_profile.sort_values(
    "missing_pct",
    ascending=False
)

print("\nTotal rows processed:", total_rows)

display(missing_profile.head(20))
```

```python
missing_profile_path = os.path.join(
    PROJECT_PATH,
    "results",
    "tables",
    "missing_value_profile.csv"
)

missing_profile.to_csv(
    missing_profile_path,
    index=False
)

print("Saved:", missing_profile_path)
```

```python
print("Missing-value distribution:")

print(
    missing_profile["missing_pct"].describe()
)
```

```python
thresholds = [50, 70, 80, 90, 95, 99]

for threshold in thresholds:
    count = (
        missing_profile["missing_pct"] >= threshold
    ).sum()

    print(
        f"Features with >= {threshold}% missing: {count}"
    )
```

```python
print("\nFeatures with < 50% missing:")

display(
    missing_profile[
        missing_profile["missing_pct"] < 50
    ].tail(20)
)
```

```python
missing_bins = pd.cut(
    missing_profile["missing_pct"],
    bins=[-0.01, 5, 20, 50, 80, 90, 100],
    labels=[
        "0-5%",
        "5-20%",
        "20-50%",
        "50-80%",
        "80-90%",
        "90-100%"
    ]
)

missing_summary = (
    missing_bins
    .value_counts()
    .sort_index()
)

print(missing_summary)
```

```python
missing_summary_df = missing_summary.reset_index()

missing_summary_df.columns = [
    "missingness_range",
    "number_of_features"
]

display(missing_summary_df)
```

```python
extreme_missing = missing_profile[
    missing_profile["missing_pct"] >= 90
].copy()

display(
    extreme_missing[
        ["feature", "missing_count", "missing_pct"]
    ]
)
```

```python
high_missing = missing_profile[
    (missing_profile["missing_pct"] >= 50) &
    (missing_profile["missing_pct"] < 90)
].copy()

display(
    high_missing[
        ["feature", "missing_count", "missing_pct"]
    ]
)
```

```python
extreme_features_path = os.path.join(
    PROJECT_PATH,
    "results",
    "tables",
    "extreme_missing_features.csv"
)

extreme_missing[
    ["feature", "missing_count", "missing_pct"]
].to_csv(
    extreme_features_path,
    index=False
)

print("Saved:", extreme_features_path)
```

```python
high_features_path = os.path.join(
    PROJECT_PATH,
    "results",
    "tables",
    "high_missing_features.csv"
)

high_missing[
    ["feature", "missing_count", "missing_pct"]
].to_csv(
    high_features_path,
    index=False
)

print("Saved:", high_features_path)
```
